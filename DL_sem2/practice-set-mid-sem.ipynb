```python
# Predict output (from mid-sem practice set)
import torch
import torch.nn as nn

model = nn.Linear(1, 1, bias=True)

with torch.no_grad():
    model.weight.fill_(1.0)
    model.bias.fill_(0.5)

optimizer = torch.optim.SGD(model.parameters(), lr=0.5)

x = torch.tensor([[2.0]])
target = torch.tensor([[3.0]])

optimizer.zero_grad()
pred = model(x)
loss = (pred - target) ** 2
loss.backward()
print("pred:", pred.item())
print("loss:", loss.item())
print("w.grad:", model.weight.grad.item())
print("b.grad:", model.bias.grad.item())
optimizer.step()
print("new_w:", model.weight.item())
print("new_b:", model.bias.item())
```

```python
import torch
import torch.nn as nn
class CustomCNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 16, kernel_size=5, stride=2, padding=1),
            nn.ReLU(),
            nn.Conv2d(16, 32, kernel_size=3, stride=1, padding=0),
            nn.ReLU(),
            nn.MaxPool2d(kernel_size=2, stride=2)
        )
        self.classifier = nn.Linear(32 * 6 * 6, 10)

    def forward(self, x):
        x = self.features(x)
        print("Feature shape:", x.shape)
        x = x.view(x.size(0), -1)
        x = self.classifier(x)
        print("Output shape:", x.shape)
        return x

model = CustomCNN()
inp = torch.randn(4, 3, 32, 32)

with torch.no_grad():
    out = model(inp)
```

```python
import torch
import torch.nn as nn
class CustomCNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(1, 32, kernel_size=5, stride=1),
            nn.MaxPool2d(2,2),
            nn.Conv2d(32, 64, kernel_size=5, stride=1),
            nn.MaxPool2d(2,2)
        )
        self.classifier = nn.Linear(64 * 7 * 7, 10)

model = CustomCNN()
inp = torch.randn(1, 1, 28, 28)

with torch.no_grad():
    out = model(inp)

print(out.shape)
```
